```python
import pandas as pd

df = pd.read_csv('StudentsPerformance.csv')
df.head()
```

```python
df['math score'] = df.fillna(df['math score'].mean())
df['reading score'] = df.fillna(df['reading score'].mean())
df['writing score'] = df.fillna(df['writing score'].mean())
```

```python
df.isna().sum()
```

```python
print("Average Marks")
print("math score average: ", df['math score'].mean())
print("reading score average: ", df['reading score'].mean())
print("writing score average: ", df['writing score'].mean())
```

```python
print("Minimmum Marks")
print("math score minimum: ", df['math score'].min())
print("reading score minimum: ", df['reading score'].min())
print("writing score minimum: ", df['writing score'].min())
```

```python
print("Maximum Marks")
print("math score maximum: ", df['math score'].max())
print("reading score maximum: ", df['reading score'].max())
print("writing score maximum: ", df['writing score'].max())
```

```python
print("Median")
print("math score median: ", df['math score'].median())
print("reading score median: ", df['reading score'].median())
print("writing score median: ", df['writing score'].median())
```

```python
print("Mode")
print("math score mode: ", df['math score'].mode())
print("reading score mode: ", df['reading score'].mode())
print("writing score mode: ", df['writing score'].mode())
```

```python
print("Sorting...")
print("Sorted by parental level of education")
df.sort_values(by='parental level of education', ascending=True).head()
```
